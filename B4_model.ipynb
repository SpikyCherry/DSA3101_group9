# feature engineering - encoding

```python
banking_marketing_train_encoded = banking_marketing_train
```

```python
# yes/no
binary_columns = ['default','housing','loan','y']
for col in binary_columns:
    print(f"Category counts for column '{col}':")
    print(banking_marketing_train[col].value_counts())
    print("\n")
for column in binary_columns:
    banking_marketing_train_encoded[column] = banking_marketing_train[column].map({'yes': 1, 'no': 0})
```

```python
# nominal
nominal_columns = ['job','marital','contact','poutcome']
for col in nominal_columns:
    print(f"Category counts for column '{col}':")
    print(banking_marketing_train[col].value_counts())
    print("\n")
banking_marketing_train_encoded = pd.get_dummies(banking_marketing_train, columns=nominal_columns)
```

```python
# create a special column for poutcome, success = 1, failure = 0, filter other columns
special_train = banking_marketing_train[banking_marketing_train['poutcome']]
```

```python
# ordinal
ordinal_columns = ['education','month']
encoder = LabelEncoder()
for col in ordinal_columns:
    print(f"Category counts for column '{col}':")
    print(banking_marketing_train[col].value_counts())
    print("\n")
    banking_marketing_train_encoded[col] = encoder.fit_transform(banking_marketing_train[col])
```

# feature engineering - create new features

## create a feature: cost
it is related to campaign, duration, contact

duration * campaign / 3600 * 38 - us dollar


from this source: https://www.magellan-solutions.com/blog/cost-of-telemarketing/

```python
banking_marketing_train.columns
```

```python
# Count the frequency of each unique value in the 'contact' column
contact_counts = banking_marketing_train['contact'].value_counts()

# Display the result
print(contact_counts)
```

```python
banking_marketing_train_encoded['cost'] = banking_marketing_train_encoded['duration'] * banking_marketing_train_encoded['campaign'] / 3600 * 38
```

```python
banking_marketing_test['cost'] = banking_marketing_test['duration'] * banking_marketing_test['campaign'] / 3600 * 38
```

```python
# Check for rows where 'contact' is 'unknown'
unknown_contact_indices = banking_marketing_train[banking_marketing_train['contact'] == 'unknown'].index

# Loop over rows with 'contact' == 'unknown'
for idx in unknown_contact_indices:
    # Ensure there's a previous row and a next row
    if idx > 0 and idx < len(banking_marketing_train) - 1:
        previous_row_cost = banking_marketing_train_encoded.loc[idx - 1, 'cost']
        next_row_cost = banking_marketing_train_encoded.loc[idx + 1, 'cost']
        
        # Calculate the average of the previous and next row's 'cost'
        average_cost = (previous_row_cost + next_row_cost) / 2
        
        # Assign the average to the current row's 'cost'
        banking_marketing_train_encoded.loc[idx, 'cost'] = average_cost

```

```python
# Check for rows where 'contact' is 'unknown'
unknown_contact_indices = banking_marketing_test[banking_marketing_test['contact'] == 'unknown'].index

# Loop over rows with 'contact' == 'unknown'
for idx in unknown_contact_indices:
    # Ensure there's a previous row and a next row
    if idx > 0 and idx < len(banking_marketing_test) - 1:
        previous_row_cost = banking_marketing_test.loc[idx - 1, 'cost']
        next_row_cost = banking_marketing_test.loc[idx + 1, 'cost']
        
        # Calculate the average of the previous and next row's 'cost'
        average_cost = (previous_row_cost + next_row_cost) / 2
        
        # Assign the average to the current row's 'cost'
        banking_marketing_test.loc[idx, 'cost'] = average_cost

```

```python
# Drop missing values from 'cost' column (if any) to ensure clean plotting
clean_cost_data = banking_marketing_train['cost'].dropna()

# Plot the distribution of the 'cost' column
sns.histplot(clean_cost_data, kde=True, color='skyblue', bins=30)

# Set plot labels and title
plt.title('Distribution of Cost')
plt.xlabel('Cost')
plt.ylabel('Density')

# Show the plot
plt.show()
```

# k means clustering for customer segmentation

```python
banking_marketing_train_encoded.columns
```

```python
banking_marketing_train_encoded['education']
```

```python
columns_for_clustering = ['age', 'education', 'default', 'balance', 'housing', 'loan', 'day',
       'month', 'duration', 'campaign', 'pdays', 'previous']
# banking_marketing_train_encoded.columns

# Extract the relevant columns
data_for_clustering = banking_marketing_train_encoded[columns_for_clustering]

# Step 3: Handle missing values (if any)
data_for_clustering = data_for_clustering.fillna(data_for_clustering.mean())

# Step 4: Standardize the data (optional but recommended for K-means)
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data_for_clustering)

# Step 5: Apply K-means clustering
# Let's choose 3 clusters for demonstration; you can experiment with this value
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(data_scaled)

# Add the cluster labels to the dataframe
banking_marketing_train_encoded['cluster'] = kmeans.labels_

# Step 6: Visualize the clusters (optional)
# You can visualize the clustering results using a pairplot or other plots
# For simplicity, we will plot the clusters using two columns (e.g., 'age' and 'balance')

plt.figure(figsize=(10, 6))
sns.scatterplot(x='age', y='balance', hue='cluster', palette='Set1', data=banking_marketing_train)
plt.title("K-means Clustering (Age vs Balance)")
plt.show()

# Step 7: Analyze the clustering results
# Check the cluster centers and distribution
print("Cluster centers:\n", kmeans.cluster_centers_)
print("\nCluster counts:\n", banking_marketing_train['cluster'].value_counts())
```
