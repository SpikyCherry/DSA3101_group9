```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os 
import sys
```

```python
current_dir = os.getcwd()
parent_dir = os.path.dirname(current_dir)
sys.path.append(parent_dir)
file_path = os.path.join(parent_dir,  "data/raw/bank_customers_train.csv")

df = pd.read_csv(file_path)
```

```python
print(df.head())
print(df.dtypes)

for col, dtype in df.dtypes.items():
    print(f"Column: {col}, Type: {dtype}")
```

```python
# Remove duplicates
print(df.duplicated().sum())
df = df.drop_duplicates() 
```

```python
# Boxplot for outliers
plt.figure(figsize=(10, 5))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
sns.barplot(x=df['marital'], y=df['y'].apply(lambda x: 1 if x == 'yes' else 0))
plt.title("Subscription Rate by Marital Status")
plt.show()
```

```python
sns.barplot(x=df['education'], y=df['y'].apply(lambda x: 1 if x == 'yes' else 0))
plt.xticks(rotation=90)
plt.title("Subscription Rate by Education Level")
plt.show()
```

```python
sns.barplot(x=df['default'], y=df['y'].apply(lambda x: 1 if x == 'yes' else 0))
sns.barplot(x=df['housing'], y=df['y'].apply(lambda x: 1 if x == 'yes' else 0))
sns.barplot(x=df['loan'], y=df['y'].apply(lambda x: 1 if x == 'yes' else 0))
plt.show()
```

```python
sns.barplot(x=df['contact'], y=df['y'].apply(lambda x: 1 if x == 'yes' else 0))
plt.title("Subscription Rate by Contact Method")
plt.show()
```

```python
sns.barplot(x=df['month'], y=df['y'].apply(lambda x: 1 if x == 'yes' else 0))
plt.xticks(rotation=90)
plt.title("Subscription Rate by Month")
plt.show()

sns.barplot(x=df['day_of_week'], y=df['y'].apply(lambda x: 1 if x == 'yes' else 0))
plt.title("Subscription Rate by Day of the Week")
plt.show()
```

```python
sns.boxplot(x=df['job'], y=df['age'], hue=df['y'])
plt.xticks(rotation=90)
plt.title("Age Distribution by Job & Subscription Status")
plt.show()
```

```python
pd.crosstab(df['education'], df['loan']).plot(kind='bar', stacked=True)
plt.title("Loan Status by Education Level")
plt.show()
```

```python
# Job vs Subscription Rate

sns.barplot(x=df['job'], y=df['y'].apply(lambda x: 1 if x == 'yes' else 0))
plt.xticks(rotation=90)
plt.title("Subscription Rate by Job Type")
plt.show()
```
